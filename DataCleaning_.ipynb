## **2: Import the required libraries: Pandas and Numpy**

## **3. Open the Excel with Data**

```python
df = pd.read_excel(r'/content/gdrive/MyDrive/predictive_maintenance_raw.xlsx', header=1)
df
```

## **4. Remove any data that isn't appropriate**
*   Any text in numeral columns
*   Any rows with empty cells
*   Any data that is not necessary
*   Any duplicates














```python
df = df.replace(['not recorded', '?', ''], np.nan)
df = df.dropna()
df = df.drop_duplicates()
df = df.drop(columns=[col for col in df.columns if "Unnamed" in col], errors='ignore')

df
```

## **5. Remove any unwanted text from data**

```python
df.dtypes
```

```python
df['temperature_F'] = df['temperature_F'].astype(str).str.replace('F','').astype(float)
df['vibration_mm_s'] = df['vibration_mm_s'].astype(str).str.replace('mm/s','').astype(float)
df['pressure_PSI'] = df['pressure_PSI'].astype(str).str.replace('PSI','').astype(float)
df['runtime_hours'] = df['runtime_hours'].astype(str).str.replace('hours','').astype(float)
df['humidity_%'] = df['humidity_%'].astype(str).str.replace('%','').astype(float)

df
```

## **6. Remove outliers**
*   Can interfere with analysis

```python
df['temperature_F'] = pd.to_numeric(df['temperature_F'])
df['vibration_mm_s'] = pd.to_numeric(df['vibration_mm_s'])
df['pressure_PSI'] = pd.to_numeric(df['pressure_PSI'])
df['runtime_hours'] = pd.to_numeric(df['runtime_hours'])
df['humidity_%'] = pd.to_numeric(df['humidity_%'])
df['failure'] = pd.to_numeric(df['failure'])
```

```python
df.loc[df['temperature_F'] <= 0, 'temperature_F'] = np.nan
df.loc[df['vibration_mm_s'] < 0, 'vibration_mm_s'] = np.nan
df.loc[df['pressure_PSI'] <= 0, 'pressure_PSI'] = np.nan
df.loc[df['runtime_hours'] < 0, 'runtime_hours'] = np.nan
df.loc[(df['humidity_%'] < 0) | (df['humidity_%'] > 100), 'humidity_%'] = np.nan

df = df.dropna()

df
```

## **7. Correct any typing errors and inconsistent naming**
*   Aluminum vs Aluminium
*   Polymr, Compos1te



```python
category_map = {
    "Temp_F": "temperature_F",
    "Vibr_mm_s": "vibration_mm_s",
    "Press_PSI": "pressure_PSI",
    "Humid%": "humidity_%",
    "Run_hours": "runtime_hours"
}

df = df.rename(columns=category_map)

df
```
