## **2. Download Libraries**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import r2_score
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

## **3. Open File**

```python
df = pd.read_excel(r'/content/gdrive/MyDrive/predictive_maintenance_raw.xlsx', header=1)
df
```

## **4. Clean column names and separate Inputs from Outputs**

```python
df = df.drop(columns=[col for col in df.columns if "Unnamed" in col], errors='ignore')

X = df.drop("failure", axis=1)
y = df["failure"]
```

## **5. Select Features and Targets**

```python
X = df[['temperature_F', 'vibration_mm_s', 'pressure_PSI', 'runtime_hours', 'humidity_%']]
y = df['failure']
```

## **7. Split Data into Training vs. Testing**

```python
X_train, X_test, y_train, y_test = train_test_split(
X, y, test_size=0.2, random_state=100)
```

## **8. Create Model**

```python
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)
```

## **9. Test how good the model is**

```python
y_pred = model.predict(X_test)
print("Predicted values:", y_pred)
```

## **10. Evaluate Performance**

```python
print("Model Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

## **11. Make a Predicition!**

```python
new_sample = pd.DataFrame({
    'temperature_F': [75.0],
    'vibration_mm_s': [3.5],
    'pressure_PSI': [35.0],
    'runtime_hours': [2000.0],
    'humidity_%': [40.0]
})

predicted_failure = model.predict(new_sample)
print("Predicted Failure:", predicted_failure[0])
```
